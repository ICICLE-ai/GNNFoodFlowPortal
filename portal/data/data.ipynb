```python
def analyze_sctg_flows(sctg_code, dest_fips=39049, top_n=10):
    # Load data
    csv_path='predicted_sctg_'+str(sctg_code)+'.csv'
    df = pd.read_csv(csv_path)

    # Filter for the given SCTG code and destination FIPS, and only existing predictions
    filtered = df[(df['dest'] == dest_fips) & (df['exist_pred'] == 1)]

    # Exclude self-loops
    filtered_no_self = filtered[filtered['origin'] != filtered['dest']]

    # Get top N origins by predicted_value_original
    top_origins = filtered_no_self.sort_values('predicted_value_original', ascending=False).head(top_n)
    origin_flows = top_origins[['origin', 'predicted_value_original']].copy()

    # Calculate total flow
    total_flow = filtered_no_self['predicted_value_original'].sum()
    origin_flows['percentage_of_total'] = (origin_flows['predicted_value_original'] / total_flow) * 100

    # Get descriptive statistics
    stats = filtered_no_self.describe()

    return origin_flows, stats
```

```python
flows, stats = analyze_sctg_flows(1)
print(flows)
```
